## 14. Find all players above or equal to level 3
Gib alle username von Spielern aus, die Level ≥ 3 erreicht haben.
Sortiere nach Level absteigend.

```python
%%sql
SELECT p.username, pl.current_level, pl.current_experience, l.experience_required
FROM PLAYER p
JOIN PLAYER_LEVEL pl ON p.id = pl.player_id
JOIN LEVEL l ON pl.current_level = l.id
WHERE pl.current_level >= 3
ORDER BY pl.current_level DESC, pl.current_experience DESC;
```

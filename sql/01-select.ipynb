Task 4: Gib mir den höchsten HP wert aus

```python
%%sql
SELECT player_id, MAX(value) AS 'Maximum Health'
FROM PLAYER_STAT
WHERE stat_id = 1
```

Task 5: Gib mir den durchschnittlichen Defense Wert von allen spielern

```python
%%sql
SELECT AVG(value) AS 'Average Defense'
FROM PLAYER_STAT
WHERE stat_id = 4
```

Task 7: Gebe mir alle Spielerwerte von Defense und Attack aus

```python
%%sql
SELECT *
FROM PLAYER_STAT
WHERE stat_id = 3 OR stat_id = 4
```

Task 8: Gib mir alle Health Werte aller Spieler aus inkl. Username

```python
%%sql
SELECT PLAYER.username, PLAYER_STAT.value AS 'Health'
FROM PLAYER
INNER JOIN PLAYER_STAT ON PLAYER.id = PLAYER_STAT.player_id
WHERE PLAYER_STAT.stat_id = 1
```

Task 9: Gib mir alle Stat werte von allen Spielern mit den Stat Namen als Spaltenüberschrift

```python
%%sql
SELECT PLAYER.username, STAT.name AS 'Stat', PLAYER_STAT.value AS 'Value'
FROM PLAYER
INNER JOIN PLAYER_STAT ON PLAYER.id = PLAYER_STAT.player_id
INNER JOIN STAT ON STAT.id = PLAYER_STAT.stat_id
```

Task 10: Gebe alle Werte von der Crit Chance alelr Spieler aus und sortiere diese von niedirg zu hoch
